```python
import pandas as pd

from sklearn.datasets import fetch_20newsgroups
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB, BernoulliNB
from sklearn.metrics import accuracy_score, f1_score
```

```python
newsgroups = fetch_20newsgroups(
    subset="all",
    remove=("headers", "footers", "quotes")
)

df = pd.DataFrame({
    "text": newsgroups.data,
    "target": newsgroups.target
})

print("Shape of dataset:", df.shape)

print("\nTarget Names:")
print(newsgroups.target_names)

print("\nFirst 5 rows:")
print(df.head())
```

```python
train_data = fetch_20newsgroups(
    subset="train",
    remove=("headers", "footers", "quotes")
)

test_data = fetch_20newsgroups(
    subset="test",
    remove=("headers", "footers", "quotes")
)

X_train = train_data.data
y_train = train_data.target

X_test = test_data.data
y_test = test_data.target

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
vectorizer = CountVectorizer(
    stop_words="english"
)

X_train_counts = vectorizer.fit_transform(X_train)
X_test_counts = vectorizer.transform(X_test)

print("Training matrix shape:", X_train_counts.shape)
print("Testing matrix shape:", X_test_counts.shape)
```

```python
multinomial_model = MultinomialNB()

multinomial_model.fit(X_train_counts, y_train)

y_pred_multinomial = multinomial_model.predict(X_test_counts)

print("Predictions generated:", len(y_pred_multinomial))
```

```python
accuracy_multinomial = accuracy_score(
    y_test,
    y_pred_multinomial
)

f1_multinomial = f1_score(
    y_test,
    y_pred_multinomial,
    average="weighted"
)

print("Multinomial Naïve Bayes")
print("Accuracy:", accuracy_multinomial)
print("F1-score:", f1_multinomial)
```

```python
binary_vectorizer = CountVectorizer(
    stop_words="english",
    binary=True
)

X_train_binary = binary_vectorizer.fit_transform(X_train)
X_test_binary = binary_vectorizer.transform(X_test)

print("Training matrix shape:", X_train_binary.shape)
print("Testing matrix shape:", X_test_binary.shape)
```

```python
bernoulli_model = BernoulliNB()

bernoulli_model.fit(X_train_binary, y_train)

y_pred_bernoulli = bernoulli_model.predict(X_test_binary)

print("Predictions generated:", len(y_pred_bernoulli))
```

```python
accuracy_bernoulli = accuracy_score(
    y_test,
    y_pred_bernoulli
)

f1_bernoulli = f1_score(
    y_test,
    y_pred_bernoulli,
    average="weighted"
)

print("Bernoulli Naïve Bayes")
print("Accuracy:", accuracy_bernoulli)
print("F1-score:", f1_bernoulli)
```

```python
comparison = pd.DataFrame({
    "Model": [
        "Multinomial Naïve Bayes",
        "Bernoulli Naïve Bayes"
    ],
    "Accuracy": [
        accuracy_multinomial,
        accuracy_bernoulli
    ],
    "F1-score": [
        f1_multinomial,
        f1_bernoulli
    ]
})

print(comparison)
```

```python
if f1_multinomial > f1_bernoulli:
    print("Multinomial Naïve Bayes performed better.")
elif f1_bernoulli > f1_multinomial:
    print("Bernoulli Naïve Bayes performed better.")
else:
    print("Both models have the same F1-score.")
```
